```python
import pandas as pd

data = {
    "Category": ["Food", "Travel", "Shopping", "Education", "Food"],
    "Amount": [200, 100, 500, 300, 150]
}

df = pd.DataFrame(data)

print("===== EXPENSE DATA =====")
print(df)

print("\n===== TOTAL EXPENSE =====")
print("₹", df["Amount"].sum())

print("\n===== CATEGORY-WISE EXPENSE =====")
print(df.groupby("Category")["Amount"].sum())
```

```python
# Find highest spending category

category_expense = df.groupby("Category")["Amount"].sum()

highest_category = category_expense.idxmax()
highest_amount = category_expense.max()

print("===== SPENDING ANALYSIS =====")
print("Highest Spending Category:", highest_category)
print("Highest Spending Amount: ₹", highest_amount)
```

```python
import pandas as pd

df = pd.read_csv("expenses.csv")

print("===== EXPENSE DATA =====")
print(df)
```

```python
# ===== EXPENSE ANALYSIS =====

print("===== SMART EXPENSE TRACKER =====")

# Total expense
total_expense = df["Amount"].sum()
print("\nTotal Expense: ₹", total_expense)

# Category-wise expense
category_expense = df.groupby("Category")["Amount"].sum()

print("\n===== CATEGORY-WISE EXPENSE =====")
print(category_expense)

# Highest spending category
highest_category = category_expense.idxmax()
highest_amount = category_expense.max()

print("\n===== HIGHEST SPENDING =====")
print("Category:", highest_category)
print("Amount: ₹", highest_amount)
```

```python
import matplotlib.pyplot as plt

# Category-wise expense
category_expense = df.groupby("Category")["Amount"].sum()

# Create bar chart
plt.figure(figsize=(8, 5))

plt.bar(category_expense.index, category_expense.values)

plt.title("Category-wise Expense Analysis")
plt.xlabel("Expense Category")
plt.ylabel("Amount (₹)")

plt.xticks(rotation=30)
plt.tight_layout()

plt.show()
```

```python
import matplotlib.pyplot as plt

category_expense = df.groupby("Category")["Amount"].sum()

plt.figure(figsize=(7, 7))

plt.pie(
    category_expense.values,
    labels=category_expense.index,
    autopct="%1.1f%%"
)

plt.title("Expense Distribution by Category")

plt.show()
```
